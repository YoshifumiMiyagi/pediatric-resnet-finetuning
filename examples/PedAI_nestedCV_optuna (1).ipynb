```python
import numpy as np
import pandas as pd
from sklearn.metrics import roc_auc_score

# ============================================================
# Settings
# ============================================================

N_BOOT = 2000
BOOT_SEED = 12345

# Nested CV OOF
oof = results_01_05["oof"].copy()

print("OOF shape:", oof.shape)
print("Columns:", oof.columns.tolist())

# ============================================================
# 1. Safety checks
# ============================================================

required = ["patient_id", "label", "oof_prob"]

missing = [c for c in required if c not in oof.columns]
if missing:
    raise ValueError(f"Missing columns: {missing}")

# 同一patient内でlabelが一致していることを確認
label_check = oof.groupby("patient_id")["label"].nunique()

assert (label_check == 1).all(), \
    "Same patient has inconsistent labels."

# OOF predictionに欠損がない
assert oof["oof_prob"].notna().all()

print("\nImages   :", len(oof))
print("Patients :", oof["patient_id"].nunique())


# ============================================================
# 2. Aggregate image predictions to patient level
#
# Multiple images / patient:
# mean OOF probability
# ============================================================

patient_oof = (
    oof
    .groupby("patient_id", as_index=False)
    .agg(
        label=("label", "first"),
        oof_prob=("oof_prob", "mean"),
        n_images=("oof_prob", "size"),
    )
)

print("\n=== Patient-level OOF ===")
print("Patients:", len(patient_oof))
print("\nLabels:")
print(patient_oof["label"].value_counts().sort_index())

print("\nImages per patient:")
print(patient_oof["n_images"].describe())


# ============================================================
# 3. Patient-level OOF AUC
# ============================================================

y_true = patient_oof["label"].to_numpy()
y_prob = patient_oof["oof_prob"].to_numpy()

patient_auc = roc_auc_score(y_true, y_prob)

print("\nPatient-level OOF AUC:")
print(f"{patient_auc:.4f}")


# ============================================================
# 4. Patient-level bootstrap 95% CI
#
# Resampling unit = patient
# ============================================================

rng = np.random.default_rng(BOOT_SEED)

n = len(patient_oof)

boot_auc = []

for _ in range(N_BOOT):

    # patient単位で復元抽出
    idx = rng.integers(
        low=0,
        high=n,
        size=n
    )

    y_b = y_true[idx]
    p_b = y_prob[idx]

    # bootstrap sampleに片方のclassしかない場合はskip
    if np.unique(y_b).size < 2:
        continue

    boot_auc.append(
        roc_auc_score(y_b, p_b)
    )

boot_auc = np.asarray(boot_auc)

ci_low, ci_high = np.percentile(
    boot_auc,
    [2.5, 97.5]
)


# ============================================================
# 5. Results
# ============================================================

print("\n" + "=" * 60)
print("PATIENT-LEVEL NESTED-CV PERFORMANCE")
print("=" * 60)

print(f"N patients     : {n}")
print(f"OOF AUC        : {patient_auc:.4f}")
print(
    f"95% bootstrap CI: "
    f"{ci_low:.4f} – {ci_high:.4f}"
)
print(f"Bootstrap reps : {len(boot_auc)}")


# ============================================================
# 6. Save
# ============================================================

patient_oof.to_csv(
    OUT_DIR / "patient_level_oof_predictions.csv",
    index=False
)

summary = pd.DataFrame({
    "age_group": ["age_01_05"],
    "n_images": [len(oof)],
    "n_patients": [n],
    "patient_oof_auc": [patient_auc],
    "ci_low": [ci_low],
    "ci_high": [ci_high],
    "n_boot": [len(boot_auc)],
})

summary.to_csv(
    OUT_DIR / "patient_level_auc_summary.csv",
    index=False
)

display(summary)
```
