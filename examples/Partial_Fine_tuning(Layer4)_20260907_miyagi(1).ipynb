```python
import numpy as np
import pandas as pd
from sklearn.metrics import roc_auc_score

# ============================================================
# Settings
# ============================================================
OOF_PATH = (
    "/content/drive/MyDrive/QC/results/"
    "test_young_layer4/oof_predictions.csv"
)

PATIENT_COL = "patient_id"
LABEL_COL = "label"

SEEDS = [42, ]
#SEEDS = [42, 43, 44, 45, 46]   # 本解析では5 seeds
N_BOOT = 2000
RANDOM_STATE = 12345


# ============================================================
# Load
# ============================================================
df = pd.read_csv(OOF_PATH)

pred_cols = [f"oof_seed{s}" for s in SEEDS]

missing = [c for c in pred_cols if c not in df.columns]
if missing:
    raise ValueError(
        f"Missing prediction columns: {missing}\n"
        "まだ1 seedしか実行していない場合は SEEDS=[42] にしてください。"
    )

print("Images:", len(df))
print("Patients:", df[PATIENT_COL].nunique())


# ============================================================
# 1. Patient-level aggregation
#    複数画像を持つ患者では予測確率を平均
# ============================================================
# 患者内でlabelが一貫しているか確認
label_check = df.groupby(PATIENT_COL)[LABEL_COL].nunique()

if (label_check > 1).any():
    bad_ids = label_check[label_check > 1].index.tolist()
    raise ValueError(
        f"同一患者内でlabelが不一致です: {bad_ids[:10]}"
    )

agg_dict = {
    LABEL_COL: "first",
}

for col in pred_cols:
    agg_dict[col] = "mean"

patient_df = (
    df.groupby(PATIENT_COL, as_index=False)
      .agg(agg_dict)
)

print("\nPatient-level dataset")
print(patient_df.shape)
print(patient_df[LABEL_COL].value_counts())


# ============================================================
# 2. Seedごとの patient-level OOF AUROC
# ============================================================
seed_results = []

y = patient_df[LABEL_COL].astype(int).to_numpy()

for seed in SEEDS:
    col = f"oof_seed{seed}"
    pred = patient_df[col].to_numpy()

    auc = roc_auc_score(y, pred)

    seed_results.append({
        "seed": seed,
        "patient_level_auc": auc
    })

seed_results = pd.DataFrame(seed_results)

mean_auc = seed_results["patient_level_auc"].mean()

# sample SD
sd_auc = seed_results["patient_level_auc"].std(ddof=1) if len(SEEDS) > 1 else np.nan


# ============================================================
# 3. Primary prediction
#    5 seedsのpatient-level OOF予測を平均
# ============================================================
patient_df["oof_mean_seeds"] = patient_df[pred_cols].mean(axis=1)

primary_pred = patient_df["oof_mean_seeds"].to_numpy()

primary_auc = roc_auc_score(y, primary_pred)


# ============================================================
# 4. Patient-level stratified bootstrap
#
#    患者単位でresampling。
#    Female / Maleをそれぞれ復元抽出してclass balanceを保持。
# ============================================================
def bootstrap_auc_patient_level(
    y_true,
    y_pred,
    n_boot=5000,
    random_state=12345
):
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)

    rng = np.random.default_rng(random_state)

    idx0 = np.where(y_true == 0)[0]
    idx1 = np.where(y_true == 1)[0]

    boot_aucs = []

    for _ in range(n_boot):

        # classごとにpatientを復元抽出
        boot0 = rng.choice(
            idx0,
            size=len(idx0),
            replace=True
        )

        boot1 = rng.choice(
            idx1,
            size=len(idx1),
            replace=True
        )

        boot_idx = np.concatenate([boot0, boot1])

        auc = roc_auc_score(
            y_true[boot_idx],
            y_pred[boot_idx]
        )

        boot_aucs.append(auc)

    boot_aucs = np.asarray(boot_aucs)

    ci_lower = np.percentile(boot_aucs, 2.5)
    ci_upper = np.percentile(boot_aucs, 97.5)

    return ci_lower, ci_upper, boot_aucs


ci_low, ci_high, boot_aucs = bootstrap_auc_patient_level(
    y,
    primary_pred,
    n_boot=N_BOOT,
    random_state=RANDOM_STATE
)


# ============================================================
# 5. Results
# ============================================================
print("\n==========================================")
print("PRIMARY ANALYSIS")
print("==========================================")

print(
    f"Patient-level AUROC = {primary_auc:.3f} "
    f"(95% CI {ci_low:.3f}-{ci_high:.3f})"
)

print("\n==========================================")
print("SEED-LEVEL REPRODUCIBILITY")
print("==========================================")

display(seed_results)

if len(SEEDS) > 1:
    print(
        f"5-seed AUROC mean ± SD = "
        f"{mean_auc:.3f} ± {sd_auc:.3f}"
    )
else:
    print(
        f"Seed {SEEDS[0]} AUROC = {mean_auc:.3f}"
    )


# ============================================================
# 6. Save
# ============================================================
summary = pd.DataFrame([{
    "n_patients": len(patient_df),
    "primary_patient_auc": primary_auc,
    "bootstrap_ci_lower": ci_low,
    "bootstrap_ci_upper": ci_high,
    "n_bootstrap": N_BOOT,
    "seed_auc_mean": mean_auc,
    "seed_auc_sd": sd_auc,
    "n_seeds": len(SEEDS),
}])

summary.to_csv(
    "/content/drive/MyDrive/QC/results/"
    "patient_level_auc_summary.csv",
    index=False
)

seed_results.to_csv(
    "/content/drive/MyDrive/QC/results/"
    "patient_level_seed_auc.csv",
    index=False
)

patient_df.to_csv(
    "/content/drive/MyDrive/QC/results/"
    "patient_level_oof_predictions.csv",
    index=False
)

print("\nSaved.")
```
